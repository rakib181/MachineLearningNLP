```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn
import requests
```

```python
url = "https://raw.githubusercontent.com/rakib181/MachineLearningNLP/refs/heads/main/Predicions_Goals.csv"
response = requests.get(url)
datasets = response.text.split('\n')
```

```python
df = pd.read_csv(url)
df.head()
```

```python
Goals = df['Goals']
Predictions = df['Predictions']

classes = {
    "Anger" : 1,
    "Fear" : 2,
    "Joy" : 3,
    "Sadness" : 4,
    "Surprise" : 5,
    "Neutral" : 6
}
y_actual = []
y_prediction = []
for i in range(len(Goals)):
    y_actual.append(classes[Goals[i]])
    y_prediction.append(classes[Predictions[i]])
```

```python
from sklearn.metrics import confusion_matrix
con_m = confusion_matrix(y_actual, y_prediction)
print(con_m)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
disp = ConfusionMatrixDisplay(
    confusion_matrix = con_m
)
disp.plot()
plt.show()
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_actual, y_prediction))
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_actual, y_prediction))
```
